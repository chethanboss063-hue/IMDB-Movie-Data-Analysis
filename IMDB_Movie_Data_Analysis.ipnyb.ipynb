# IMDb Movie Data Analysis

### A Data Analysis Project Using Python

## Introduction

Movies generate a large amount of data related to ratings, genres, directors, actors, runtime, votes, revenue, and Metascore. This project analyzes an IMDb movie dataset using Python to discover useful patterns and insights.

The project focuses on data cleaning, exploratory data analysis, statistical analysis, and data visualization. Different Python libraries such as Pandas, Matplotlib, and Seaborn are used to process and visualize the data.

## Objectives

- To understand and explore the IMDb movie dataset.
- To clean and preprocess the dataset.
- To identify missing and duplicate values.
- To analyze movie ratings, revenue, votes, runtime, and genres.
- To identify highly rated and popular movies.
- To analyze relationships between different movie attributes.
- To create meaningful visualizations from the data.
- To derive useful insights from the dataset.

## Methodology

The project follows a systematic data analysis process:

1. **Data Collection**  
   The IMDb movie dataset was obtained from the project repository.

2. **Data Loading**  
   The CSV dataset was loaded into Google Colab using Python and Pandas.

3. **Data Exploration**  
   The dataset was examined using functions such as `head()`, `tail()`, `shape()`, `info()`, and `describe()`.

4. **Data Cleaning**  
   Missing values were identified using `isnull().sum()`. Records containing missing values were removed using the `dropna()` method.

5. **Duplicate Checking**  
   Duplicate records were checked using the `duplicated()` method.

6. **Data Analysis**  
   Movies were analyzed based on rating, genre, revenue, votes, runtime, year, directors, actors, and Metascore.

7. **Additional Analysis**  
   New analyses were performed to identify top-rated movies, highly voted movies, highest-revenue movies, frequent actors, genre characteristics, and relationships between movie attributes.

8. **Data Visualization**  
   Charts and plots were created using Matplotlib and Seaborn to represent important findings visually.

9. **Interpretation**  
   The results were examined to identify meaningful patterns and insights from the IMDb dataset.

# Additional Analysis

# Additional Analysis of IMDB movie dataset

---



## 22. Display Top 10 Movies Based on Number of Votes

```python
top_votes = data.nlargest(10, 'Votes')[['Title', 'Votes']].set_index('Title')

top_votes
```

## 23. Top 10 Directors by Number of Movies

```python
top_directors = data['Director'].value_counts().head(10)

top_directors
```

## 24. Average IMDb Rating by Genre

```python
genre_rating = data.groupby('Genre')['Rating'].mean().sort_values(ascending=False)

genre_rating
```

## 25. Top 10 Movies with Highest Revenue

```python
top_revenue = data.nlargest(10, 'Revenue (Millions)')[['Title', 'Revenue (Millions)']].set_index('Title')

top_revenue
```

## 26. Movies Released After 2010 with Rating Above 8

```python
high_rated_recent = data[
    (data['Year'] > 2010) &
    (data['Rating'] > 8)
][['Title', 'Year', 'Rating']].sort_values(
    by='Rating',
    ascending=False
)

high_rated_recent
```

## 27. Relationship Between Rating and Revenue

```python
correlation = data['Rating'].corr(data['Revenue (Millions)'])

print("Correlation between Rating and Revenue:", round(correlation, 3))
```

## 28. Top 10 Most-Voted Movies

```python
top_voted = data.nlargest(10, 'Votes')[['Title', 'Votes', 'Rating']]

top_voted
```

## 29. Top 10 Movies by Revenue - Visualization

```python
plt.figure(figsize=(10, 6))

plt.barh(
    top_revenue['Revenue (Millions)'].index,
    top_revenue['Revenue (Millions)']
)

plt.xlabel('Revenue (Millions)')
plt.ylabel('Movie Title')
plt.title('Top 10 Movies by Revenue')

plt.gca().invert_yaxis()
plt.xticks(rotation=45)

plt.show()
```

## 30. Top 10 Movies by IMDb Rating - Visualization

```python
top_rated = data.nlargest(10, 'Rating')[['Title', 'Rating']]

plt.figure(figsize=(10, 6))

plt.barh(
    top_rated['Title'],
    top_rated['Rating']
)

plt.xlabel('IMDb Rating')
plt.ylabel('Movie Title')
plt.title('Top 10 Highest-Rated Movies')

plt.gca().invert_yaxis()
plt.xlim(0, 10)

plt.show()
```

## 31. Relationship Between IMDb Rating and Revenue - Visualization

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    data['Rating'],
    data['Revenue (Millions)'],
    alpha=0.6
)

plt.xlabel('IMDb Rating')
plt.ylabel('Revenue (Millions)')
plt.title('Relationship Between IMDb Rating and Revenue')

plt.show()
```

## 32. Top 10 Most Frequent Actors
